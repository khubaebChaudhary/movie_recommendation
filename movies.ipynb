```python
import nltk
from nltk.stem.porter import PorterStemmer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
```

```python
nltk.download('punkt')
```

```python
# Initialize stemmer
ps = PorterStemmer()
```

cosidering words like love lover lovers as one

```python
# Apply stemming
def stem_words(text):
   return " ".join([ps.stem(word) for word in text.split()])

# Apply stemming to tags
new_df['tags'] = new_df['tags'].apply(stem_words)
```

```python
#  Using TF-IDF for better text representation (captures relative importance)
tfidf = TfidfVectorizer(max_features=5000, stop_words='english')
vector = tfidf.fit_transform(new_df['tags']).toarray()
```

```python
def recommend(movie, top_n=5):
    try:
        # Convert input to lowercase
        movie = movie.lower()
        
        # Find index of the movie
        index = new_df[new_df['title'] == movie].index[0]
        
        # Compute similarity of this movie to all others (lazy computation)
        movie_vector = vector[index].reshape(1, -1)
        sim_scores = cosine_similarity(movie_vector, vector)[0]
        
        # Get top N similar movie indices (excluding itself)
        similar_indices = np.argsort(sim_scores)[::-1][1:top_n+1]
        
        print(f"\n🎬 Top {top_n} movies similar to '{movie.title()}':")
        for i in similar_indices:
            print(" -", new_df.iloc[i]['title'].title())  # Capitalize titles for output

    except IndexError:
        print(f"Movie '{movie}' not found in the database.")
```

```python
recommend('Interstellar')
```
